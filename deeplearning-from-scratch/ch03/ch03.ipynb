### 理解 NumPy 数组的 `ndim` 和 `shape` 的关键在于**轴（axis）**的概念。每个数组可以看作是一个 **N 维网格**，`ndim` 告诉你这个网格有多少个轴（维度），`shape` 告诉你每个轴上的元素个数。

---

### 1. 一维数组 `A = np.array([1, 2, 3, 4])`

- **`ndim` = 1**：只有一条轴（轴 0）。可以想象成一根直线上排列着 4 个点。
- **`shape` = (4, )**：元组里只有一个数字，表示轴 0 的长度为 4。括号后面的逗号是 Python 元组的写法，表示它是一个单元素元组。

你可以这样理解：
- 用 **一个下标** 就能访问到元素：`A[0]` → 1，`A[3]` → 4。
- 没有“行”或“列”的概念，就是一列数值（但横着写还是竖着写在显示上不影响本质）。

---

### 2. 二维数组 `B = np.array([[1,2], [3,4], [5,6]])`

- **`ndim` = 2**：有两条轴：轴 0（垂直方向，常称为“行”），轴 1（水平方向，常称为“列”）。
- **`shape` = (3, 2)**：轴 0 长度为 3（有 3 行），轴 1 长度为 2（有 2 列）。

访问元素需要 **两个下标**：`B[0, 1]` → 第一行第二列的值 2；`B[2, 0]` → 第三行第一列的值 5。

你可以把二维数组想象成一个表格（矩阵）：
```
行索引↓  列索引→ 0   1
        0   1   2
        1   3   4
        2   5   6
```

---

### 类比扩展

- **标量**（单个数值）：`ndim = 0`，`shape = ()`（空元组），没有轴。
- **三维数组**：例如 `shape = (2, 3, 4)` 表示 2 层，每层 3 行 4 列。需要三个下标 `[z, y, x]` 访问。可以想象成一个长方体。

---

### 关键要点

- `ndim` 就是 `len(shape)`，即 shape 元组的长度。
- 维度数（ndim）就是你需要多少个整数下标才能唯一确定数组中的一个元素。
- 形状（shape）告诉你每个下标允许的取值范围（从 0 到该维度长度-1）。

这种表示方式是 NumPy 和深度学习框架（如 PyTorch、TensorFlow）处理张量的基础，理解它对后续数组操作、广播、矩阵乘法等非常关键。

### 问：轴 0（垂直方向，常称为“行”），轴 1（水平方向，常称为“列”），行不是水平方向吗？为什么说是垂直，怎么理解？

你这个问题提得很好，确实是容易混淆的地方。关键在于区分 **“行本身的排列方向”** 和 **“沿着哪个轴变化”**。

### 1. 直观理解：行的视觉排列 vs 索引变化的方向

- **行（row）** 在视觉上确实是 **水平** 排列的：一行里有多个元素，从左到右。
- **轴 0（axis=0）** 在二维数组中对应的是 **行的索引**。当你改变轴 0 的下标（比如从第 0 行变到第 1 行），你的视线是在 **垂直方向** 上下移动，而不是水平方向。

举个例子，考虑数组 `B`：

```
行0 →  [1, 2]
行1 →  [3, 4]
行2 →  [5, 6]
```

- 固定轴 1（列索引）不变，沿着轴 0 走动：取第 0 列的所有元素 `[1, 3, 5]` —— 这是 **垂直向下** 走的。
- 所以 **轴 0 对应垂直方向**（自上而下），它决定你选哪一行。

### 2. 用 NumPy 操作来验证

```python
import numpy as np
B = np.array([[1,2], [3,4], [5,6]])
print(B.sum(axis=0))   # 沿着轴0求和 -> [9, 12] (把每一列从上到下加起来)
print(B.sum(axis=1))   # 沿着轴1求和 -> [3, 7, 11] (把每一行从左到右加起来)
```

- `axis=0` 把同一列的不同行加起来 → 垂直方向计算。
- `axis=1` 把同一行的不同列加起来 → 水平方向计算。

### 3. 为什么说“轴0是行，轴1是列”？

这是指 **轴0的大小是行数，轴1的大小是列数**。但“行”这个名词暗示了水平排列，这与“垂直方向”似乎矛盾。
其实不矛盾：**“行”是一个对象（水平的一排数据）**，**“沿着轴0移动”是在不同行对象之间切换**，这种切换在物理布局上是垂直的。就像书架：每层书是水平排列的（行），但你在不同层之间上下移动（轴0）。

### 4. 更通用的记忆法（适用于任意维度）

- **轴 n** 就是 **第 n 个下标** 变化的方向。
- 对于二维数组：
  - 下标 `[i, j]` 中 `i` 对应轴 0，`i` 变化 → 上下移动。
  - `j` 对应轴 1，`j` 变化 → 左右移动。

所以总结成一句话：
> **二维数组中，轴 0 是垂直方向（行索引变化），轴 1 是水平方向（列索引变化）。**
> 行本身是水平排列的，但“轴0”指的是索引行号的那个维度，它在空间上对应垂直方向。

````python
def init_network():
    network = {}
    network['W1'] = np.array([[0.1, 0.3, 0.5], [0.2, 0.4, 0.6]])
    network['b1'] = np.array([0.1, 0.2, 0.3])
    network['W2'] = np.array([[0.1, 0.4], [0.2, 0.5], [0.3, 0.6]])
    network['b2'] = np.array([0.1, 0.2])
    network['W3'] = np.array([[0.1, 0.3], [0.2, 0.4]])
    network['b3'] = np.array([0.1, 0.2])
    return network
````

````python
def forward(network, x):
    W1, W2, W3 = network['W1'], network['W2'], network['W3']
    b1, b2, b3 = network['b1'], network['b2'], network['b3']

    a1 = np.dot(x, W1) + b1
    z1 = sigmoid(a1)
    a2 = np.dot(z1, W2) + b2
    z2 = sigmoid(a2)
    a3 = np.dot(z2, W3) + b3
    y = identity_function(a3)

    return y
````

````python
network = init_network()
x = np.array([1.0, 0.5])
y = forward(network, x)
print(y) # [ 0.31682708 0.69627909]
````

````python
def softmax(a):
    exp_a = np.exp(a)
    sum_exp_a = np.sum(exp_a)
    y = exp_a / sum_exp_a

    return y
````

````python
def softmax(a):
    c = np.max(a)
    exp_a = np.exp(a - c) # 溢出对策
    sum_exp_a = np.sum(exp_a)
    y = exp_a / sum_exp_a

    return y
````

````python
import pickle

def get_data():
    (x_train, t_train), (x_test, t_test) = \
        load_mnist(normalize=True, flatten=True, one_hot_label=False)
    return x_test, t_test

def init_network():
    with open("sample_weight.pkl", 'rb') as f:
        network = pickle.load(f)

    return network

def predict(network, x):
    W1, W2, W3 = network['W1'], network['W2'], network['W3']
    b1, b2, b3 = network['b1'], network['b2'], network['b3']
    a1 = np.dot(x, W1) + b1
    z1 = sigmoid(a1)
    a2 = np.dot(z1, W2) + b2
    z2 = sigmoid(a2)
    a3 = np.dot(z2, W3) + b3
    y = softmax(a3)

    return y

x, t = get_data()
network = init_network()

x, _ = get_data()
network = init_network()
W1, W2, W3 = network['W1'], network['W2'], network['W3']
print(x.shape)
print(x[0].shape)
print(W1.shape)
print(W2.shape)
print(W3.shape)

accuracy_cnt = 0
for i in range(len(x)):
    y = predict(network, x[i])
    p = np.argmax(y) # 获取概率最高的元素的索引
    if p == t[i]:
        accuracy_cnt += 1

print("Accuracy:" + str(float(accuracy_cnt) / len(x)))
````

### 怎么理解 np.argmax(x, axis=1)

`np.argmax(x, axis=1)` 的意思是：**沿着轴 1（水平方向/列方向）寻找最大值的位置**，对每一行独立操作，返回该行最大值所在的列索引（0, 1 或 2）。

### 具体到你的例子

```python
x = np.array([[0.1, 0.8, 0.1],   # 行0
              [0.3, 0.1, 0.6],   # 行1
              [0.2, 0.5, 0.3],   # 行2
              [0.8, 0.1, 0.1]])  # 行3
y = np.argmax(x, axis=1)
```

逐行看：
- **行0**：`[0.1, 0.8, 0.1]` 最大值是 `0.8`，位于第 1 列 → 索引 `1`
- **行1**：`[0.3, 0.1, 0.6]` 最大值是 `0.6`，位于第 2 列 → 索引 `2`
- **行2**：`[0.2, 0.5, 0.3]` 最大值是 `0.5`，位于第 1 列 → 索引 `1`
- **行3**：`[0.8, 0.1, 0.1]` 最大值是 `0.8`，位于第 0 列 → 索引 `0`

所以 `y` 的结果是：
```python
array([1, 2, 1, 0])
```

### 如何形象理解 axis=1

- 把 `x` 想象成一个表格，有 4 行 3 列。
- `axis=0` 对应 **从上到下**（跨行）→ 对每一列找最大值索引。
- `axis=1` 对应 **从左到右**（跨列）→ 对每一行找最大值索引。

你可以记住口诀：**axis 表示要“压扁”的维度**。`axis=1` 意味着沿着列的方向滑动，把这一行里的所有列比较一遍，每行留下一个最大值的位置。最终结果比原数组少一维（从二维变成了一维）。

````python
x, t = get_data()
network = init_network()

batch_size = 100 # 批数量
accuracy_cnt = 0

for i in range(0, len(x), batch_size):
    x_batch = x[i:i+batch_size]
    y_batch = predict(network, x_batch)
    p = np.argmax(y_batch, axis=1)
    accuracy_cnt += np.sum(p == t[i:i+batch_size])

print("Accuracy:" + str(float(accuracy_cnt) / len(x)))
````
